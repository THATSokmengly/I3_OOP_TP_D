### Principle class of OOP 
In OOP python programming, we  know that python has 4 main Principle class that including: Absraction, Encapsulation, Inheritance and Polymorphism that it Using in  different purpose.

### Encapsulation: 
Encapsulation means keeping data and the methods that work with that data together inside a class. It helps protect data and maintain its accuracy.

```python
class Fish:

    def __init__(self):
        self.__size = "big"

    def get_size(self):
        print("I'm a " + self.__size + " fish")

    def set_size(self, new_size):
        self.__size = new_size 

oscar = Fish()
oscar.get_size() 

bert = Fish()
bert.__size = "small" 
bert.get_size() 

fin = Fish()
fin.set_size("tiny")
fin.get_size() 
```

### Abstraction
Abstraction means hiding unnecessary implementation details and showing only the important features of an object. It allows users to understand and use an object without knowing how it works internally.

```python
class Dog:

    def __init__(self, name):
        self.name = name 
        print(self.name + " was adopted.")

    def bark(self):
        print("woof!")

spot = Dog("spot") 
spot.bark() 
```

### Inheritance
Inheritance allows a new class (child class) to receive attributes and methods from an existing class (parent class). It helps reuse existing code and reduces code duplication.

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

dog1 = Dog()
dog1.speak()
```

### Polymorphism
Polymorphism means that different classes can use the same method name but perform different actions. It allows different objects to share a common interface while having their own behaviors.

```python
class Dog:
    def speak(self):
        print("Dogbarks")

class Cat:
    def speak(self):
        print("Catmeows")

animals= [Dog(), Cat()]
for animal in animals:
    animal.speak()
```

```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self._price = price  # Protected attribute
   
    @property
    def price(self):
        return self._price
    @price.setter
    def price(self, new_price):
        if new_price > 0:
            self._price = new_price
        else:
            print("Price must be greater than 0!")
   
    def display_info(self):
        print("Book Information")
        print("--------------------")
        print("Title:", self.title)
        print("Author:", self.author)
        print("Price: $", self.price)

class EBook(Book):
    def __init__(self, title, author, price, file_size):
        # Inherit attributes from Book
        super().__init__(title, author, price)
        self.file_size = file_size
    # Additional method for EBook
    def display_ebook_info(self):
        self.display_info()
        print("File Size:", self.file_size, "MB")

book1 = Book("Python Basics", "John Smith", 25)
# Display book information
book1.display_info()
print("\n********************\n")

print("Original Price:", book1.price)

book1.price = 30
print("Updated Price:", book1.price)

book1.price = -10
print("\n********************\n")

ebook1 = EBook(
    "Python Programming",
    "David",
    15,
    50
)
ebook1.display_ebook_info()

```
